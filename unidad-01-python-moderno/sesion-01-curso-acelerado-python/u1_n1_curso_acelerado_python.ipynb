### Mutabilidad, alias y copia

Dos nombres pueden apuntar a la misma lista. Para modificar el contenedor por separado, debemos copiarlo.
Si contiene otros objetos mutables, una copia superficial aún puede compartirlos. Este detalle explica muchos errores silenciosos
en pipelines de datos.

```python
original = ["dato-a", "dato-b"]
alias = original
copia = original.copy()

alias.append("dato-c")
copia.append("dato-independiente")

print("original:", original)
print("alias:   ", alias)
print("copia:   ", copia)
print("¿Mismo objeto?", original is alias, original is copia)
```

### Copiar una estructura anidada

`.copy()` hace una **copia superficial**: crea el contenedor externo, pero conserva
las referencias a sus elementos. Si dentro hay una lista, esa lista sigue compartida.

```python
registro_original = {"texto": "hola", "etiquetas": ["saludo"]}
registro_copia = registro_original.copy()
registro_copia["etiquetas"].append("corto")
print(registro_original)
print(registro_original is registro_copia)
print(registro_original["etiquetas"] is registro_copia["etiquetas"])
```

## 3. Decisiones y truthiness

### Bloques, ciclos y decisiones

Los dos puntos abren un bloque y la sangría indica qué instrucciones pertenecen a
él. Usa cuatro espacios por nivel. `for` repite el bloque para cada elemento;
`range(3)` produce los valores 0, 1 y 2. `continue` salta al siguiente elemento y
`break` termina el ciclo. `while` repite mientras una condición sea verdadera.

```python
for valor in [0.0, None, 0.85]:
    if valor is None:
        continue
    print("Score presente:", valor)

intentos = 0
while intentos < 3:
    intentos += 1
    print("Intento", intentos)
```

## 4. Ciclos y comprehensions

Un `for` recorre los elementos de un iterable. En una lista puedes recorrer los
valores directamente; no hace falta construir sus índices. Usa `enumerate` cuando
también necesites la posición y `zip` para avanzar por varias secuencias a la vez.
Por defecto, `zip` se detiene al agotarse la secuencia más corta. Si necesitas
comprobar que tengan la misma longitud, usa `zip(..., strict=True)`.

```python
textos = ["muy bueno", "regular", "excelente"]
scores = [0.82, 0.51, 0.96]

for posicion, (texto, score) in enumerate(zip(textos, scores), start=1):
    print(f"{posicion}. {texto:<12} → {score:.0%}")
```

Las comprehensions sirven para transformaciones pequeñas y legibles. Si la
lógica requiere varias decisiones o efectos secundarios, un ciclo explícito
suele ser mejor.

```python
scores_altos = [score for score in scores if score >= 0.80]
longitudes = {texto: len(texto) for texto in textos}
iniciales = {texto[0] for texto in textos}

print(scores_altos)
print(longitudes)
print(iniciales)
```

### Ejercicio 6 · Transformar y filtrar

Convierte a Fahrenheit las temperaturas mayores o iguales a 25. Usa la fórmula `fahrenheit = celsius * 9 / 5 + 32`. Primero usa un ciclo y después una comprehension. El resultado debe ser `[77.0, 84.2, 87.8]`.

```python
temperaturas = [18, 21, 25, 29, 31]
# Construye ambas versiones.

# Versión con ciclos
temperaturas_fahrenheit = []
for temperatura in temperaturas:
    if temperatura >= 25:
        fahrenheit = temperatura * 9 / 5 + 32
        temperaturas_fahrenheit.append(fahrenheit)
print(temperaturas_fahrenheit)

# Versión con comprensión
temperaturas_fahrenheit_comp = [
    temperatura * 9 / 5 + 32 for temperatura in temperaturas if temperatura >= 25
]
print(temperaturas_fahrenheit_comp)
```

## 5. Funciones: nombrar una transformación

Una función bien enfocada reduce repetición y permite probar una idea de forma
aislada. Sus parámetros son entradas; `return` produce una salida. Una función
sin `return` explícito devuelve `None`.

```python
def normalizar_texto(texto, *, minusculas=True):
    '''Elimina espacios externos y opcionalmente convierte a minúsculas.'''
    resultado = texto.strip()
    if minusculas:
        resultado = resultado.lower()
    return resultado


print(normalizar_texto("  Hola, IA  "))
print(normalizar_texto("  Hola, IA  ", minusculas=False))
```

El `*` vuelve explícitos los argumentos posteriores: debemos escribir
`minusculas=False`. En la semana 2 añadiremos anotaciones de tipo y Pydantic;
hoy nos concentramos en el comportamiento.

```python
assert normalizar_texto("  HOLA  ") == "hola"
assert normalizar_texto("  HOLA  ", minusculas=False) == "HOLA"
assert normalizar_texto("") == ""
```

### Ejercicio 7 · Promedio sin inventar datos

Escribe `promedio_o_none(valores)`: devuelve el promedio o `None` si la lista está vacía. Comprueba `[0.0]`, `[]` y `[0.5, 1.0]`. ¿Por qué cero y ausencia necesitan resultados distintos?

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
def promedio_o_none(valores):
    if not valores:
        return None
    return sum(valores) / len(valores)

# Comprobaciones requeridas
assert promedio_o_none([0.0]) == 0.0
assert promedio_o_none([]) is None
assert promedio_o_none([0.5, 1.0]) == 0.75

# Cero (0.0) representa una métrica real calculada a partir de observaciones
# válidas.
# Ausencia (None) representa un estado indeterminado debido a la falta de
# mediciones, previniendo errores de división por cero.
```

### Ordenar usando una función

`sorted` crea una lista nueva; `.sort()` modifica la lista existente. El argumento
`key` recibe una función que calcula el criterio de orden. Pasamos la función
sin llamarla: Python la aplicará a cada elemento.

```python
evaluaciones = [
    {"modelo": "a", "score": 0.7},
    {"modelo": "b", "score": 0.95},
    {"modelo": "c", "score": 0.8},
]

def obtener_score(evaluacion):
    return evaluacion["score"]

ranking = sorted(evaluaciones, key=obtener_score, reverse=True)
print(ranking)
# lambda permite expresar una función pequeña en el mismo lugar.
assert ranking == sorted(evaluaciones, key=lambda e: e["score"], reverse=True)
```

### De contar a agrupar

Un contador responde «¿cuántos?». Una agrupación conserva los elementos asociados
a cada clave. `defaultdict(list)` crea una lista al acceder a una clave ausente.

```python
from collections import defaultdict

conteo_manual = {}
for etiqueta in etiquetas:
    conteo_manual[etiqueta] = conteo_manual.get(etiqueta, 0) + 1
assert conteo_manual == dict(Counter(etiquetas))

por_etiqueta = defaultdict(list)
for registro in [
    {"etiqueta": "positivo", "score": 0.9},
    {"etiqueta": "neutral", "score": 0.6},
    {"etiqueta": "positivo", "score": 0.8},
]:
    por_etiqueta[registro["etiqueta"]].append(registro["score"])
print(dict(por_etiqueta))
```

### Ejercicio 8 · Resumir grupos

## 8. POO: objetos que conservan estado

Una **clase** define cómo construir objetos y qué operaciones ofrecen. Cada
**instancia** es un objeto concreto. Sus **atributos** guardan datos y sus
**métodos** son funciones asociadas al objeto.

`__init__` inicializa una instancia; `self` representa esa instancia dentro de los
métodos. Al llamar `contador.agregar()`, Python pasa `contador` como `self`.

```python
class Contador:
    def __init__(self, nombre):
        self.nombre = nombre
        self.total = 0

    def agregar(self):
        self.total += 1

    def reiniciar(self):
        self.total = 0

contador_a = Contador("lote A")
contador_b = Contador("lote B")
contador_a.agregar()
contador_a.agregar()
print(contador_a.nombre, contador_a.total)
print(contador_b.nombre, contador_b.total)
assert contador_b.total == 0
```

### Laboratorio POO: una lista compartida por accidente

Un atributo definido en el cuerpo de la clase se comparte entre las instancias.
Los datos propios de cada instancia deben inicializarse en `__init__`.

```python
class RegistroCompartido:
    registros = []  # Defecto intencional: la lista pertenece a la clase.

    def agregar(self, texto):
        self.registros.append(texto)

compartido_a = RegistroCompartido()
compartido_b = RegistroCompartido()
compartido_a.agregar("dato de A")
print(compartido_b.registros)
assert compartido_a.registros is compartido_b.registros
```

## 9. Caso integrador: un registro de predicciones

La clase siguiente conserva predicciones ya normalizadas y combina estructuras de
datos con comportamiento y estado. Cada registro recibe su posición al agregarse.
`resumen` devuelve cantidad, conteos y promedio; el promedio es `None` cuando no
hay registros.


```python
predicciones = [
    {"texto": "Excelente", "etiqueta": "positivo", "score": 0.98},
    {"texto": "Regular", "etiqueta": "neutral", "score": 0.66},
    {"texto": "Llegó roto", "etiqueta": "negativo", "score": 0.93},
    {"texto": "Cero válido", "etiqueta": "neutral", "score": 0.0},
]
```

```python
from collections import Counter, defaultdict

class RegistroPredicciones:
    def __init__(self, nombre):
        self.nombre = nombre
        self._registros = []

    def agregar(self, registro, posicion):
        guardado = registro.copy()
        guardado["posicion"] = posicion
        self._registros.append(guardado)

    def resumen(self):
        scores = [registro["score"] for registro in self._registros]
        return {
            "cantidad": len(self._registros),
            "conteos": dict(Counter(r["etiqueta"] for r in self._registros)),
            "promedio": sum(scores) / len(scores) if scores else None,
        }

    def promedio_por_etiqueta(self):
        # Agrupa los scores por cada etiqueta única
        grupos = defaultdict(list)
        for r in self._registros:
            grupos[r["etiqueta"]].append(r["score"])

        # Calcula el promedio de cada grupo
        return {
            etiqueta: sum(scores) / len(scores)
            for etiqueta, scores in grupos.items()
        }

    def reiniciar(self):
        # Vacía el listado interno de registros
        self._registros.clear()

lote = RegistroPredicciones("Demostración")
for posicion, registro in enumerate(predicciones):
    lote.agregar(registro, posicion)
print(lote.resumen())
```

y promedio **0.6425**. Este promedio describe los scores; no mide exactitud del
modelo porque no tenemos etiquetas de referencia.

## Respuestas y Conclusiones del Taller

### 1. Cuándo elegir cada estructura de datos
* **Lista de diccionarios** (`predicciones`): Es ideal para almacenar colecciones de registros donde el orden de llegada importa (cronología) y cada registro cuenta con múltiples atributos con nombres específicos (`texto`, `etiqueta`, `score`).
* **Conjunto** (`set`): Es la mejor opción para extraer valores únicos (como las etiquetas distintas de un lote) de forma eficiente y realizar pruebas rápidas de pertenencia (`in`) o álgebra de conjuntos (uniones, intersecciones).

### 2. Diferencia entre cero y ausencia
* **Cero (`0.0`)** es un número y representa una métrica real válida calculada a partir de los datos (un score muy bajo o neutral).
* **Ausencia (`None`)** indica una falta total de información o que el cálculo no es posible en este momento (un lote vacío). Confundir ambos puede llevar a divisiones por cero (`ZeroDivisionError`) o a sesgar un promedio de manera incorrecta.

### 3. Copiar contenedores con elementos mutables
* Al usar `.copy()`, se genera una **copia superficial**. Esto significa que si el contenedor original tiene listas o diccionarios anidados en su interior, la copia compartirá las mismas referencias de memoria de esos objetos anidados. Si modificamos un elemento interno mutable en la copia, la modificación se reflejará también en el objeto original. Para evitarlo, se debe copiar individualmente cada nivel anidado (copia manual) o utilizar `copy.deepcopy()`.

### 4. Anatomía de `RegistroPredicciones` y Estado
* **Atributos:**
  * `nombre` (Atributo de instancia público para identificar el lote).
  * `_registros` (Atributo de instancia privado/protegido por convención con guion bajo, que actúa como memoria interna).
* **Métodos:**
  * `__init__(self, nombre)`: Inicializa el estado.
  * `agregar(self, registro, posicion)`: Añade registros asegurando el aislamiento mediante `.copy()`.
  * `resumen(self)`: Calcula estadísticas agregadas.
  * `promedio_por_etiqueta(self)`: Agrupa y promedia usando `defaultdict`.
  * `reiniciar(self)`: Limpia el estado interno.
* **¿Por qué deben conservar listas distintas?**
  Si los lotes compartieran la misma lista (por ejemplo, si estuviera declarada a nivel de clase fuera del `__init__`), añadir un registro a un lote afectaría a todos los demás de manera colateral. El estado independiente en `__init__` garantiza que cada instancia de evaluación sea un entorno aislado y seguro.

### 5. Función vs. Objeto: ¿Cuándo ayuda conservar el estado?
* **Función pura (sin estado):** Es util para flujos síncronos sencillos donde disponemos de todo el conjunto de datos de antemano (un dataframe cargado completo) y se realiza una transformación rápida de una sola vez.
* **Instancia con estado:** Indispensable cuando los datos llegan de forma incremental o asíncrona a lo largo del tiempo (por ejemplo, en sistemas en tiempo real, streaming o microservicios donde las predicciones se registran de una en una mediante peticiones http). El objeto mantiene la memoria acumulada facilitando la entrega de resúmenes en cualquier momento sin obligar a la API externa a reenviar todo el historial.
